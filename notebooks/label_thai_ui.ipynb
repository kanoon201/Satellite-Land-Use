# A1: label Thai patches by eye

Every labeller works alone and looks at all 200 patches, then picks the class they see. The answers of the
whole team are voted on afterwards, and the patches people disagree on are discussed together.
The WorldCover label and the area name are **not shown** unless you press the reveal button (see below), so they
do not steer your answer. Do not compare answers with the others until everybody has finished.

**Before you start**
1. In the second code cell set `LABELER` (your name, no spaces, different from everybody else's).
   Leave `GROUP = "all"` to label all 200 patches; `GROUP = 1`, `2` or `3` gives only that group's 60 patches plus the 20 shared ones (80 in total).
2. Run the cells from top to bottom. The last cell shows one picture: the reference examples on the left, the patch to label on the right,
   and the answer buttons under it. Everything is on one screen.
3. If the picture is too big or too small for your screen, change `FIG_WIDTH` in the second code cell and run the last cell again.

**Buttons**

| button | meaning |
|---|---|
| 1 forest, 2 crop, 3 grassland, 4 built-up, 5 water | the class you see |
| 0 unsure | none of these, or two classes are equally likely (bare soil, shrub, cloud, ...) |
| undo | removes the last answer, so you can answer that patch again |
| show WorldCover label | shows what WorldCover says about this patch. Use it only when you are unsure; the answer is then saved with `peeked = True`, so it can be counted separately |

To stop, just close the notebook: every click is saved at once. Run the last cell again later to continue where you left off.
Your answers go to `outputs/manual_labels/<LABELER>.csv`. Commit only your own file.

**Draft guidelines** (the team should agree on these before labelling; edit this cell if you change them)

- **forest:** dense tree canopy, dark green, rough texture, no field edges
- **crop:** fields with straight edges and regular shapes, green, bare or stubble
- **grassland:** smooth low vegetation without field pattern and without trees
- **built-up:** buildings, roads, regular street grids; a patch that is mostly roofs and pavement
- **water:** smooth, uniform, dark or blue surface
- **unsure:** use it when two classes are equally likely; do not guess. Unsure answers are counted separately.

Look at both pictures: true colour shows what the eye would see; in false colour vegetation is red and water is almost black.
